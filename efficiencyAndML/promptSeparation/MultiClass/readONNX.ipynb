```python
import onnx
```

```python
FILE = "output/MultiClass/modelBDT_pt0_100.onnx"

model = onnx.load(FILE)

print("Model:")
print(model.graph.name)
```

```python
print("\nInputs:")
for x in model.graph.input:
    print(x.name, ":\n", x.type.tensor_type.shape)
```

```python
print("\nOutputs:")
for x in model.graph.output:
    print(x.name, ":\n", x.type.tensor_type.shape)
```

```python
print("\nNodes:", len(model.graph.node))
```

```python

onnx.checker.check_model(model)
print("\nModel is valid.")
```
